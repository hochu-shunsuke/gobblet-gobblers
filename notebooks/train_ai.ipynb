## Step 4: データ収集（1000ゲーム）

```
# 設定
N_GAMES = 500  # 片側500ゲーム（計1000ゲーム）

minimax = MinimaxPlayer(max_depth=4)
random_ai = RandomPlayer()

print("=" * 50)
print("📊 Data Collection for AI Training")
print("=" * 50)

all_records = []

# 1. Minimax（先手） vs Random
print(f"\n🎮 Collecting: Minimax vs Random ({N_GAMES} games)...")
start = time.time()
arena1 = Arena(minimax, random_ai, verbose=False)
for i in range(N_GAMES):
    record = arena1.play_game()
    all_records.append(record)
    if (i+1) % 100 == 0:
        print(f"  {i+1}/{N_GAMES} completed...")
print(f"  ✅ Done in {time.time()-start:.1f}s")

# 2. Random vs Minimax（後手）
print(f"\n🎮 Collecting: Random vs Minimax ({N_GAMES} games)...")
start = time.time()
arena2 = Arena(random_ai, minimax, verbose=False)
for i in range(N_GAMES):
    record = arena2.play_game()
    all_records.append(record)
    if (i+1) % 100 == 0:
        print(f"  {i+1}/{N_GAMES} completed...")
print(f"  ✅ Done in {time.time()-start:.1f}s")

# 統計
minimax_wins = sum(1 for r in all_records if 
    (r['player1'].startswith('Minimax') and r['winner'] == 1) or
    (r['player2'].startswith('Minimax') and r['winner'] == 2))

print(f"\n📈 Summary:")
print(f"  Total games: {len(all_records)}")
print(f"  Minimax win rate: {minimax_wins/len(all_records)*100:.1f}%")
```

## Step 5: データ保存 & ダウンロード

```
# データ保存
filename = f"training_data_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
with open(filename, 'w') as f:
    json.dump(all_records, f)
print(f"💾 Saved to: {filename}")

# ファイルサイズ確認
import os
size_mb = os.path.getsize(filename) / (1024 * 1024)
print(f"📦 File size: {size_mb:.2f} MB")

# Colabからダウンロード
from google.colab import files
files.download(filename)
```
